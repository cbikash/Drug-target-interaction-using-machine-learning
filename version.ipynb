```python
import sys
import platform

import numpy as np
import sklearn
import torch
import xgboost as xgb
import rdkit
import shap



# SYSTEM / SOFTWARE CONFIGURATION
if torch.backends.mps.is_available():
    hardware = "Apple Silicon / MPS and CPU"
elif torch.cuda.is_available():
    hardware = f"CUDA GPU ({torch.cuda.get_device_name(0)}) and CPU"
else:
    hardware = "CPU"


configuration = {
    "Python": platform.python_version(),
    "XGBoost": xgb.__version__,
    "NumPy": np.__version__,
    "Deep-learning framework": f"PyTorch {torch.__version__}",
    "Molecular toolkit": f"RDKit {rdkit.__version__}",
    "Protein model": "ESM-2",
    "Baseline ML library": f"scikit-learn {sklearn.__version__}",
    "Explainability": f"SHAP {shap.__version__}",
    "Primary hardware": hardware,
}


# PRINT AS MARKDOWN TABLE
print("| Component | Configuration |")

for component, value in configuration.items():
    print(f"| {component} | {value} |")
```
