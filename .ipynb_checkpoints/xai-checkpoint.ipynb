```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
from captum.attr import IntegratedGradients, Saliency, GuidedBackprop, LayerGradCam, LayerAttribution
from captum.attr import visualization as viz
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA
import pandas as pd

class ExplainableDTICNN(nn.Module):
    def __init__(self, ligand_dim=1024, protein_dim=320, conv_channels=[64, 128], 
                 fc_dims=[256, 128, 1], dropout_rate=0.3, l2_lambda=1e-4):
        super(ExplainableDTICNN, self).__init__()
        
        self.l2_lambda = l2_lambda
        
        # Store intermediate layers for XAI
        self.conv1 = nn.Conv1d(1, conv_channels[0], kernel_size=5, padding=2)
        self.bn1 = nn.BatchNorm1d(conv_channels[0])
        self.relu1 = nn.ReLU()
        self.dropout1 = nn.Dropout(dropout_rate)
        
        self.conv2 = nn.Conv1d(conv_channels[0], conv_channels[1], kernel_size=3, padding=1)
        self.bn2 = nn.BatchNorm1d(conv_channels[1])
        self.relu2 = nn.ReLU()
        self.dropout2 = nn.Dropout(dropout_rate)
        
        # Store for GradCAM
        self.ligand_conv_features = None
        
        # Protein processing
        self.protein_fc1 = nn.Linear(protein_dim, conv_channels[0])
        self.protein_bn1 = nn.BatchNorm1d(conv_channels[0])
        self.protein_relu1 = nn.ReLU()
        self.protein_dropout1 = nn.Dropout(dropout_rate)
        
        self.protein_fc2 = nn.Linear(conv_channels[0], conv_channels[1])
        self.protein_bn2 = nn.BatchNorm1d(conv_channels[1])
        self.protein_relu2 = nn.ReLU()
        self.protein_dropout2 = nn.Dropout(dropout_rate)
        
        # Combined fully connected layers
        ligand_flat_dim = conv_channels[1] * 512  # After conv + pooling
        protein_flat_dim = conv_channels[1]
        
        self.fc1 = nn.Linear(ligand_flat_dim + protein_flat_dim, fc_dims[0])
        self.fc1_bn = nn.BatchNorm1d(fc_dims[0])
        self.fc1_relu = nn.ReLU()
        self.fc1_dropout = nn.Dropout(dropout_rate)
        
        self.fc2 = nn.Linear(fc_dims[0], fc_dims[1])
        self.fc2_bn = nn.BatchNorm1d(fc_dims[1])
        self.fc2_relu = nn.ReLU()
        self.fc2_dropout = nn.Dropout(dropout_rate)
        
        self.output = nn.Linear(fc_dims[1], fc_dims[2])
        
        self._initialize_weights()
    
    def _initialize_weights(self):
        for m in self.modules():
            if isinstance(m, nn.Linear) or isinstance(m, nn.Conv1d):
                nn.init.xavier_uniform_(m.weight, gain=0.1)
                if m.bias is not None:
                    nn.init.constant_(m.bias, 0)
    
    def forward(self, ligand_features, protein_features):
        # Handle tensor dimensions
        if ligand_features.dim() == 4:
            ligand_features = ligand_features.squeeze(1).squeeze(1)
        elif ligand_features.dim() == 3 and ligand_features.size(1) == 1:
            ligand_features = ligand_features.squeeze(1)
        
        # Ligand processing with hook for GradCAM
        x = ligand_features.unsqueeze(1)  # [batch, 1, 1024]
        x = self.conv1(x)
        x = self.bn1(x)
        x = self.relu1(x)
        x = self.dropout1(x)
        
        x = self.conv2(x)
        x = self.bn2(x)
        x = self.relu2(x)
        self.ligand_conv_features = x  # Store for GradCAM
        x = self.dropout2(x)
        
        lig_x = x.view(x.size(0), -1)  # Flatten
        
        # Protein processing
        prot_x = self.protein_fc1(protein_features)
        prot_x = self.protein_bn1(prot_x)
        prot_x = self.protein_relu1(prot_x)
        prot_x = self.protein_dropout1(prot_x)
        
        prot_x = self.protein_fc2(prot_x)
        prot_x = self.protein_bn2(prot_x)
        prot_x = self.protein_relu2(prot_x)
        prot_x = self.protein_dropout2(prot_x)
        
        # Combine features
        combined = torch.cat((lig_x, prot_x), dim=1)
        
        # Final layers
        x = self.fc1(combined)
        x = self.fc1_bn(x)
        x = self.fc1_relu(x)
        x = self.fc1_dropout(x)
        
        x = self.fc2(x)
        x = self.fc2_bn(x)
        x = self.fc2_relu(x)
        x = self.fc2_dropout(x)
        
        output = self.output(x)
        return output

class XAIAnalyzer:
    def __init__(self, model, device):
        self.model = model
        self.device = device
        
        # Initialize attribution methods
        self.integrated_gradients = IntegratedGradients(self.model)
        self.saliency = Saliency(self.model)
        self.guided_backprop = GuidedBackprop(self.model)
        
        # For GradCAM - attach to the last conv layer
        self.gradcam = LayerGradCam(
            self.model, 
            self.model.ligand_conv_features
        )
    
    def explain_predictions(self, ligand_input, protein_input, target_class=None):
        """
        Generate explanations for model predictions
        """
        # Move inputs to device
        ligand_input = ligand_input.to(self.device)
        protein_input = protein_input.to(self.device)
        
        # Ensure proper dimensions
        if ligand_input.dim() == 4:
            ligand_input = ligand_input.squeeze(1).squeeze(1)
        elif ligand_input.dim() == 3 and ligand_input.size(1) == 1:
            ligand_input = ligand_input.squeeze(1)
        
        # Add batch dimension if missing
        if ligand_input.dim() == 1:
            ligand_input = ligand_input.unsqueeze(0)
        if protein_input.dim() == 1:
            protein_input = protein_input.unsqueeze(0)
        
        # Forward pass to get predictions
        self.model.eval()
        with torch.no_grad():
            predictions = self.model(ligand_input, protein_input)
        
        explanations = {}
        
        # 1. Integrated Gradients for both ligand and protein
        lig_attr_ig, prot_attr_ig = self._compute_integrated_gradients(
            ligand_input, protein_input, target_class
        )
        explanations['integrated_gradients'] = {
            'ligand': lig_attr_ig,
            'protein': prot_attr_ig,
            'predictions': predictions
        }
        
        # 2. Saliency maps
        lig_attr_sal, prot_attr_sal = self._compute_saliency(
            ligand_input, protein_input, target_class
        )
        explanations['saliency'] = {
            'ligand': lig_attr_sal,
            'protein': prot_attr_sal
        }
        
        # 3. Guided Backpropagation
        lig_attr_gbp, prot_attr_gbp = self._compute_guided_backprop(
            ligand_input, protein_input, target_class
        )
        explanations['guided_backprop'] = {
            'ligand': lig_attr_gbp,
            'protein': prot_attr_gbp
        }
        
        return explanations, predictions
    
    def _compute_integrated_gradients(self, ligand_input, protein_input, target_class=None):
        """Compute Integrated Gradients attribution"""
        baseline_lig = torch.zeros_like(ligand_input)
        baseline_prot = torch.zeros_like(protein_input)
        
        lig_attr, prot_attr = self.integrated_gradients.attribute(
            inputs=(ligand_input, protein_input),
            baselines=(baseline_lig, baseline_prot),
            target=target_class,
            return_convergence_delta=True
        )
        
        return lig_attr[0].cpu().detach().numpy(), prot_attr[0].cpu().detach().numpy()
    
    def _compute_saliency(self, ligand_input, protein_input, target_class=None):
        """Compute Saliency attribution"""
        lig_attr, prot_attr = self.saliency.attribute(
            inputs=(ligand_input, protein_input),
            target=target_class
        )
        
        return lig_attr.cpu().detach().numpy(), prot_attr.cpu().detach().numpy()
    
    def _compute_guided_backprop(self, ligand_input, protein_input, target_class=None):
        """Compute Guided Backpropagation attribution"""
        lig_attr, prot_attr = self.guided_backprop.attribute(
            inputs=(ligand_input, protein_input),
            target=target_class
        )
        
        return lig_attr[0].cpu().detach().numpy(), prot_attr[0].cpu().detach().numpy()
    
    def visualize_explanations(self, explanations, sample_idx=0):
        """Visualize explanation results"""
        fig, axes = plt.subplots(2, 3, figsize=(18, 12))
        
        methods = ['integrated_gradients', 'saliency', 'guided_backprop']
        titles = ['Integrated Gradients', 'Saliency', 'Guided Backprop']
        
        for i, (method, title) in enumerate(zip(methods, titles)):
            exp_data = explanations[method]
            
            # Plot ligand attributions
            axes[0, i].plot(exp_data['ligand'][sample_idx])
            axes[0, i].set_title(f'{title} - Ligand ({method})')
            axes[0, i].set_xlabel('Feature Index')
            axes[0, i].set_ylabel('Attribution Score')
            axes[0, i].grid(True, alpha=0.3)
            
            # Plot protein attributions
            axes[1, i].bar(range(len(exp_data['protein'][sample_idx])), 
                          exp_data['protein'][sample_idx])
            axes[1, i].set_title(f'{title} - Protein ({method})')
            axes[1, i].set_xlabel('Feature Index')
            axes[1, i].set_ylabel('Attribution Score')
            axes[1, i].tick_params(axis='x', rotation=45)
        
        plt.tight_layout()
        plt.show()
    
    def feature_importance_analysis(self, explanations):
        """Analyze feature importance across different methods"""
        results = {}
        
        for method, exp_data in explanations.items():
            # Ligand feature importance
            lig_imp = np.abs(exp_data['ligand']).mean(axis=0)
            prot_imp = np.abs(exp_data['protein']).mean(axis=0)
            
            results[method] = {
                'ligand_importance': lig_imp,
                'protein_importance': prot_imp,
                'top_ligand_features': np.argsort(lig_imp)[-10:],  # Top 10 most important
                'top_protein_features': np.argsort(prot_imp)[-10:]
            }
        
        return results
    
    def generate_feature_rankings(self, explanations):
        """Generate ranked feature lists for interpretation"""
        feature_ranks = {}
        
        for method, exp_data in explanations.items():
            # Rank ligand features
            lig_attr = exp_data['ligand']
            lig_mean_attr = np.abs(lig_attr).mean(axis=0)
            lig_ranked = np.argsort(lig_mean_attr)[::-1]  # Descending order
            
            # Rank protein features
            prot_attr = exp_data['protein']
            prot_mean_attr = np.abs(prot_attr).mean(axis=0)
            prot_ranked = np.argsort(prot_mean_attr)[::-1]  # Descending order
            
            feature_ranks[method] = {
                'ligand_rankings': lig_ranked,
                'protein_rankings': prot_ranked
            }
        
        return feature_ranks
    
    def plot_top_features(self, explanations, n_features=20):
        """Plot top contributing features"""
        fig, axes = plt.subplots(1, 2, figsize=(16, 6))
        
        # Get top features from integrated gradients (most reliable method)
        ig_data = explanations['integrated_gradients']
        
        # Top ligand features
        lig_attr = np.abs(ig_data['ligand']).mean(axis=0)
        top_lig_idx = np.argsort(lig_attr)[-n_features:]
        top_lig_values = lig_attr[top_lig_idx]
        
        axes[0].barh(range(len(top_lig_idx)), top_lig_values)
        axes[0].set_yticks(range(len(top_lig_idx)))
        axes[0].set_yticklabels([f'Feature {i}' for i in top_lig_idx])
        axes[0].set_xlabel('Average Attribution Score')
        axes[0].set_title(f'Top {n_features} Ligand Features')
        axes[0].grid(True, alpha=0.3)
        
        # Top protein features
        prot_attr = np.abs(ig_data['protein']).mean(axis=0)
        top_prot_idx = np.argsort(prot_attr)[-n_features:]
        top_prot_values = prot_attr[top_prot_idx]
        
        axes[1].barh(range(len(top_prot_idx)), top_prot_values)
        axes[1].set_yticks(range(len(top_prot_idx)))
        axes[1].set_yticklabels([f'Feature {i}' for i in top_prot_idx])
        axes[1].set_xlabel('Average Attribution Score')
        axes[1].set_title(f'Top {n_features} Protein Features')
        axes[1].grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.show()
    
    def correlation_analysis(self, explanations):
        """Analyze correlations between different explanation methods"""
        methods = list(explanations.keys())
        n_methods = len(methods)
        
        # Prepare correlation matrix
        correlations = np.zeros((n_methods, n_methods))
        
        for i, method1 in enumerate(methods):
            for j, method2 in enumerate(methods):
                # Calculate correlation for ligand features
                attr1 = explanations[method1]['ligand'].flatten()
                attr2 = explanations[method2]['ligand'].flatten()
                corr = np.corrcoef(attr1, attr2)[0, 1]
                correlations[i, j] = corr
        
        # Plot correlation heatmap
        plt.figure(figsize=(10, 8))
        sns.heatmap(correlations, annot=True, cmap='coolwarm', center=0,
                   xticklabels=methods, yticklabels=methods,
                   vmin=-1, vmax=1)
        plt.title('Correlation Matrix - Different XAI Methods (Ligand Features)')
        plt.tight_layout()
        plt.show()
        
        return correlations

def demonstrate_xai_usage(model, test_loader, device, n_samples=5):
    """Demonstrate XAI analysis on test samples"""
    xai_analyzer = XAIAnalyzer(model, device)
    
    # Get a few samples from test loader
    samples = []
    for i, (ligand_batch, protein_batch, target_batch) in enumerate(test_loader):
        if i >= n_samples:
            break
        samples.append((ligand_batch, protein_batch, target_batch))
    
    all_explanations = []
    all_predictions = []
    
    for i, (ligand_batch, protein_batch, target_batch) in enumerate(samples):
        print(f"\nAnalyzing sample {i+1}/{len(samples)}...")
        
        # Generate explanations
        explanations, predictions = xai_analyzer.explain_predictions(
            ligand_batch[:1],  # Take first sample from batch
            protein_batch[:1],
            target_class=None
        )
        
        all_explanations.append(explanations)
        all_predictions.append(predictions)
        
        print(f"Prediction: {predictions.item():.4f}, Target: {target_batch[0].item():.4f}")
    
    # Analyze all explanations
    print("\n=== FEATURE IMPORTANCE ANALYSIS ===")
    feature_importance = xai_analyzer.feature_importance_analysis(all_explanations[0])
    
    for method, imp_data in feature_importance.items():
        print(f"\n{method.upper()} - Top 5 Ligand Features:")
        top_lig = imp_data['top_ligand_features']
        print([f"Feature {idx}: {imp_data['ligand_importance'][idx]:.4f}" for idx in top_lig[-5:]])
        
        print(f"\n{method.upper()} - Top 5 Protein Features:")
        top_prot = imp_data['top_protein_features']
        print([f"Feature {idx}: {imp_data['protein_importance'][idx]:.4f}" for idx in top_prot[-5:]])
    
    # Visualize explanations for first sample
    print("\n=== VISUALIZING EXPLANATIONS ===")
    xai_analyzer.visualize_explanations(all_explanations[0])
    
    # Plot top features
    xai_analyzer.plot_top_features(all_explanations[0])
    
    # Correlation analysis
    correlations = xai_analyzer.correlation_analysis(all_explanations[0])
    
    return xai_analyzer, all_explanations

# Example usage with your existing model
def integrate_xai_with_training():
    # Assuming you have your trained model
    device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    
    # Load your existing model or create new one
    model = ExplainableDTICNN(
        ligand_dim=1024,
        protein_dim=320,
        conv_channels=[64, 128],
        fc_dims=[256, 128, 1],
        dropout_rate=0.3
    ).to(device)
    
    # Load your trained weights if available
    try:
        checkpoint = torch.load('best_regularized_model.pth', map_location=device)
        model.load_state_dict(checkpoint['model_state_dict'])
        print("Loaded trained model weights")
    except:
        print("No saved model found, using untrained model")
    
    # Create some example data for demonstration
    # In practice, you would use your test data
    example_ligand = torch.randn(1, 1024).to(device)
    example_protein = torch.randn(1, 320).to(device)
    
    # Initialize XAI analyzer
    xai_analyzer = XAIAnalyzer(model, device)
    
    # Generate explanations
    explanations, predictions = xai_analyzer.explain_predictions(
        example_ligand, example_protein
    )
    
    print(f"Prediction: {predictions.item():.4f}")
    
    # Show feature importance
    feature_ranks = xai_analyzer.generate_feature_rankings(explanations)
    
    # Display top features for each method
    for method, ranks in feature_ranks.items():
        print(f"\n{method.upper()} - Top 10 Ligand Features:")
        top_lig = ranks['ligand_rankings'][:10]
        print(top_lig)
        
        print(f"{method.upper()} - Top 10 Protein Features:")
        top_prot = ranks['protein_rankings'][:10]
        print(top_prot)
    
    # Visualize the explanations
    xai_analyzer.visualize_explanations(explanations)
    
    return xai_analyzer, explanations

if __name__ == "__main__":
    # Run XAI integration
    xai_analyzer, explanations = integrate_xai_with_training()
    
    print("\nXAI analysis completed! The explanations show which molecular features")
    print("are most important for the model's predictions, helping interpret")
    print("the drug-target interaction predictions.")
```
